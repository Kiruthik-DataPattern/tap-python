```
import csv

source = "archive/olist_orders_dataset.csv"
target = "archive/olist_orders_semicolon.csv"

with open(source, newline="", encoding="utf-8") as infile:
    reader = csv.DictReader(infile)
    columns = reader.fieldnames

    with open(target, "w", newline="", encoding="utf-8") as outfile:
        writer = csv.DictWriter(outfile, fieldnames=columns, delimiter=";")
        writer.writeheader()

        for row in reader:
            writer.writerow(row)

print("Copy created:", target)
```

```
def count_rows(path, delimiter):
    with open(path, newline="", encoding="utf-8") as file:
        reader = csv.DictReader(file, delimiter=delimiter)
        count = 0
        for row in reader:
            count += 1
        return count

print(count_rows("archive/olist_orders_dataset.csv", ","))
print(count_rows("archive/olist_orders_semicolon.csv", ";"))
```
